```python
from qpsychometric.questionnaire_validator import QuestionnaireValidator
from qpsychometric.personality_traits.big5 import big5_questionnaire
```

```python
# Import big5 questions
big5_qmnli_df = big5_questionnaire['QMNLI']
big5_questions = big5_qmnli_df.get_questions()

big5_factors = list(set([q()._descriptor['Factor'] for q in big5_questions]))
```

```python
# Create validator instance
validator = QuestionnaireValidator(
    questionnaire_name="big5",
    questions=big5_questions,
    factors=big5_factors,
    result_path="results/",
)

results = validator.run_validation()
```

```python
validator = QuestionnaireValidator(
    questionnaire_name="Big5",
    questions=big5_questions,
    factors=big5_factors,
    result_path="results/",
)

# Returns a DataFrame for all questions — no MNLI models needed
semantic_df = validator.get_semantic_similarity()   # q=None → all questions
cola_df = validator.get_cola_score()                # q=None → all questions

# Merge into a single content validity view
import pandas as pd
content_validity = semantic_df.join(cola_df)
content_validity
```
