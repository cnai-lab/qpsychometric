```python
from qpsychometric.questionnaire_validator import QuestionnaireValidator
from qpsychometric.mental_health.sense_of_coherence import soc_questionnaire
```

```python
# Import soc questions
soc_qmnli_df = soc_questionnaire['QMNLI']
soc_questions = soc_qmnli_df.get_questions()

soc_factors = list(set([q()._descriptor['Factor'] for q in soc_questions]))
```

```python
# Create validator instance
validator = QuestionnaireValidator(
    questionnaire_name="soc",
    questions=soc_questions,
    factors=soc_factors,
    result_path="results/",
)

results = validator.run_validation()
```
