```python
from qpsychometric.questionnaire_validator import QuestionnaireValidator
from qpsychometric.mental_health.patient_health_questionnaire import phq_questionnaire
```

```python
# Import phq questions
phq_qmnli_df = phq_questionnaire['QMNLI']
phq_questions = phq_qmnli_df.get_questions()

phq_factors = list(set([q()._descriptor['Factor'] for q in phq_questions]))
```

```python
# Create validator instance
validator = QuestionnaireValidator(
    questionnaire_name="phq",
    questions=phq_questions,
    factors=phq_factors,
    result_path="results/",
)

results = validator.run_validation()
```
