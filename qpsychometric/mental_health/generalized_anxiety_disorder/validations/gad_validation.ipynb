```python
from qpsychometric.questionnaire_validator import QuestionnaireValidator
from qpsychometric.mental_health.generalized_anxiety_disorder import gad_questionnaire
```

```python
# Import gad questions
gad_qmnli_df = gad_questionnaire['QMNLI']
gad_questions = gad_qmnli_df.get_questions()

gad_factors = list(set([q()._descriptor['Factor'] for q in gad_questions]))
```

```python
# Create validator instance
validator = QuestionnaireValidator(
    questionnaire_name="gad",
    questions=gad_questions,
    factors=gad_factors,
    result_path="results/",
)

results = validator.run_validation()
```
