```python
from qpsychometric.questionnaire_clm_validator import QuestionnaireCLMValidator
from qpsychometric.social_biases.ambivalent_sexism_inventory import asi_questionnaire
```

```python
# Import asi questions
asi_qmnli_df = asi_questionnaire['QCLM']
asi_questions = asi_qmnli_df.get_questions()

asi_factors = list(set([q()._descriptor['Factor'] for q in asi_questions]))
```

```python
# Create validator instance
validator = QuestionnaireCLMValidator(
    questionnaire_name="asi",
    questions=asi_questions,
    factors=asi_factors,
    result_path="results/",
    update=True
)
```

```python
results = validator.run_validation()
```

```python
results.keys()
```

```python
results_csv = results['results_csv']
```

## Analysis - base models

```python
content_validity_base = validator.calc_content_validity(results_csv, model_type='base')
content_validity_base
```

```python
alpha_base = validator.calc_cronbach_alpha(results_csv, model_type='base')
print('overall:', alpha_base['overall'])
print('factors:', alpha_base['factors'])
```

```python
validator.test_question_affect_on_cronbach_alpha(alpha_base['data_df'])
```

```python
correlations_base = validator.calc_correlations(results_csv, model_type='base')
correlations_base
```
